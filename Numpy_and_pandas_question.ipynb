```python
import numpy as np
```

```python
arr = np.arange(11, 30, 2)
print(arr)
print(arr.ndim)
print(arr.size)
```

```python
arr = np.linspace(5, 35, 7)
print(arr)
```

```python
arr = np.full((4, 3), 9)
arr[2, 1] = 0
print(arr)
```

```python
a = np.array([6, 12, 18, 24])
print(a % 5)
```

```python
arr = np.array([14, 28, 42, 56, 70, 84])
print(arr[1:6:2])
```

```python
arr = np.array([[7, 8],
                [9, 10],
                [11, 12]])
transposed = arr.T
print(transposed)
```
