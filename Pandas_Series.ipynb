```python
import pandas as pd
from pandas import Series
```

```python
sales = pd.Series([120, 150, 180, 210])
print(sales)
```

```python
marks = Series([72, 81, 69, 90])
print(marks.values)
```

```python
temperatures = pd.Series(
    [28, 31, 30, 29],
    index=["Mon", "Tue", "Wed", "Thu"]
)

print(temperatures)
```

```python
population = Series(
    [35, 42, 31],
    index=["Delhi", "Mumbai", "Jaipur"]
)

print(population["Jaipur"])
```

```python
score = Series([35, 68, 44, 79, 52])

print(score[score > 50])
```

```python
subjects = Series(
    [76, 83, 91],
    index=["Physics", "Chemistry", "Maths"]
)

print("Chemistry" in subjects.index)
```

```python
s = Series([10, 20, 30], index=["A", "B", "C"])

print(s.to_dict())
```

```python
prices = {
    "Pen": 10,
    "Notebook": 55,
    "Bag": 700
}

price_series = pd.Series(prices)

print(price_series)
```

```python
pd.Series({"Pen": 10, "Bag": 700})
```

```python
data = {"A": 10, "B": 20, "C": 30}

s = Series(data, index=["A", "B", "C", "D"])

print(s)
```

```python
print(s.notnull())
print(s.isnull())
```

```python
s = Series([500, 600, 700], index=["East", "West", "North"])

print("West" in s)
print(500 in s)
```

```python
s = Series([18, 45, 27, 61], index=["P", "Q", "R", "S"])

print(s[s > 30])
```

```python
city_data = {
    "Delhi": 80,
    "Pune": 65,
    "Kochi": 70
}

s = Series(
    city_data,
    index=["Delhi", "Kochi", "Pune", "Chennai"]
)

print(s)
```

```python
s = Series(
    {"W": 5, "X": 10, "Y": 15},
    index=["W", "X", "Y", "Z"]
)

print(pd.isnull(s))
```

```python
print(pd.notnull(s))
```

```python
s1 = Series([10, 20, 30], index=["A", "B", "C"])
s2 = Series([1, 2, 3], index=["C", "A", "B"])

print(s1 + s2)
```

```python
s1 = Series({"A": 12, "B": 18})
s2 = Series({"B": 5, "C": 9})

print(s1 + s2)
```

```python
stock = Series([25, 40, 18], index=["Item1", "Item2", "Item3"])

stock_dict = stock.to_dict()
stock2 = Series(stock_dict)

print("Dictionary:", stock_dict)
print("New Series:")
print(stock2)
```

```python
marks = {
    "Maths": 88,
    "Science": 74,
    "English": 69,
    "History": 81
}

s = Series(
    marks,
    index=["History", "English", "Science", "Maths"]
)

print("Original Series:")
print(s)

print("Marks greater than 70:")
print(s[s > 70])
```

```python
north = Series({"Jan": 40, "Feb": 55, "Mar": 60})
south = Series({"Feb": 20, "Mar": 25, "Apr": 30})

total = north + south

print(total)
```

```python
d = {"Red": 12, "Blue": 18, "Green": 24}

labels = ["Green", "Red", "Yellow", "Blue"]

result = Series(d, index=labels)

print(result)
print(pd.isnull(result))
```

```python
s = Series([90, 75, 82], index=["Ravi", "Aman", "Neha"])

d = s.to_dict()

s2 = Series(d)

print("Dictionary:", d)
print("s2:")
print(s2)
```
